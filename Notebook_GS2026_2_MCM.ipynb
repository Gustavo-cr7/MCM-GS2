## ETAPA 1 (5 pontos) — Diagnóstico e Modelagem da Situação Atual
### a) Identificação das Variáveis
- $x_1$ ($N$): Quantidade atual de ônibus alocados na região **Norte** ($x_1 \in \mathbb{Z}_+$)
- $x_2$ ($S$): Quantidade atual de ônibus alocados na região **Sul** ($x_2 \in \mathbb{Z}_+$)
- $x_3$ ($L$): Quantidade atual de ônibus alocados na região **Leste** ($x_3 \in \mathbb{Z}_+$)
- $x_4$ ($O$): Quantidade atual de ônibus alocados na região **Oeste** ($x_4 \in \mathbb{Z}_+$)

### b) Modelo Matemático (Sistema Linear)
$$\begin{cases} 1x_1 + 1x_2 + 1x_3 + 1x_4 = 72 \\ 1x_1 + 2x_2 + 0x_3 + 1x_4 = 76 \\ 0x_1 + 1x_2 + 2x_3 + 0x_4 = 52 \\ 1x_1 + 0x_2 + 1x_3 + 2x_4 = 76 \end{cases}$$

### c) Representação Matricial ($A\mathbf{x} = \mathbf{b}$)
$$\begin{bmatrix} 1 & 1 & 1 & 1 \\ 1 & 2 & 0 & 1 \\ 0 & 1 & 2 & 0 \\ 1 & 0 & 1 & 2 \end{bmatrix} \begin{bmatrix} x_1 \\ x_2 \\ x_3 \\ x_4 \end{bmatrix} = \begin{bmatrix} 72 \\ 76 \\ 52 \\ 76 \end{bmatrix}$$

### d) Investigação Matemática de Existência e Unicidade (Álgebra Linear)
- **Determinante e Invertibilidade:** $\det(A) = 3 \neq 0$, logo $A$ é invertível ($\exists A^{-1}$) e suas colunas são linearmente independentes (LI).
- **Teorema de Rouché-Capelli:** $\text{posto}(A) = \text{posto}([A\mid\mathbf{b}]) = n = 4$, logo o sistema é **Possível e Determinado (SPD)** e admite solução única $\mathbf{x} = A^{-1}\mathbf{b} = (12, 20, 16, 24)^T$.

```python
#etapa 1
import numpy as np

A = np.array([
    [1, 1, 1, 1],
    [1, 2, 0, 1],
    [0, 1, 2, 0],
    [1, 0, 1, 2]
])

B = np.array([72, 76, 52, 76])

det = np.linalg.det(A)
posto_A = np.linalg.matrix_rank(A)
posto_AB = np.linalg.matrix_rank(np.column_stack((A, B)))

print("Determinante:", det)
print(f"Posto(A): {posto_A} | Posto([A|B]): {posto_AB} | Incógnitas n: {A.shape[1]}")

if not np.isclose(det, 0):
    X = np.linalg.solve(A, B)

    print("\nDistribuição atual dos ônibus:")
    regioes = ["Norte", "Sul", "Leste", "Oeste"]

    for regiao, quantidade in zip(regioes, X):
        print(f"{regiao}: {quantidade:.0f} ônibus")

    print("\nVerificação das equações:")
    print(A @ X)
else:
    print("O sistema não possui solução única.")
```

## ETAPA 2 (5 pontos) — Desenvolvimento de uma Política de Redistribuição
### a) Estimativa de Ônibus Necessários e b) Identificação de Desequilíbrios
- **Norte:** $560 / 40 = 14$ necessários vs. $12$ disponíveis $\implies$ **Déficit de $-2$ ônibus** ($-80$ vagas)
- **Sul:** $760 / 40 = 19$ necessários vs. $20$ disponíveis $\implies$ **Superávit de $+1$ ônibus** ($+40$ vagas)
- **Leste:** $720 / 40 = 18$ necessários vs. $16$ disponíveis $\implies$ **Déficit de $-2$ ônibus** ($-80$ vagas)
- **Oeste:** $840 / 40 = 21$ necessários vs. $24$ disponíveis $\implies$ **Superávit de $+3$ ônibus** ($+120$ vagas)

### c, d, e) Política de Redistribuição, Transformação Linear $\mathbf{y} = T\mathbf{x}$ e Matriz Associada
Adotamos a política de **compartilhamento proporcional à demanda** ($t_{ij} = n_i / 72$), representada pela transformação linear $T: \mathbb{R}^4 \to \mathbb{R}^4$ com matriz associada:
$$T = \frac{1}{72}\begin{bmatrix} 14 & 14 & 14 & 14 \\ 19 & 19 & 19 & 19 \\ 18 & 18 & 18 & 18 \\ 21 & 21 & 21 & 21 \end{bmatrix}$$
Cada coeficiente $t_{ij}$ representa a fração da frota do centro $j$ destinada à região $i$.

### f) Demonstração de Conservação da Frota Total e Não Negatividade
1. **Conservação da frota:** A matriz $T$ é **estocástica por colunas** ($\sum_{i=1}^4 t_{ij} = 1$ para todo $j$, ou seja, $\mathbf{1}^T T = \mathbf{1}^T$). Assim, $\mathbf{1}^T \mathbf{y} = \mathbf{1}^T (T\mathbf{x}) = (\mathbf{1}^T T)\mathbf{x} = \mathbf{1}^T \mathbf{x} = 72$.
2. **Não negatividade:** Como $t_{ij} \ge 0$ e $x_j \ge 0$, temos $y_i = \sum_{j=1}^4 t_{ij} x_j \ge 0$ para todo $i$.

```python
#etapa 2
import numpy as np

# Distribuição atual encontrada na Etapa 1
frota_atual = np.linalg.solve(
    np.array([
        [1, 1, 1, 1],
        [1, 2, 0, 1],
        [0, 1, 2, 0],
        [1, 0, 1, 2]
    ]),
    np.array([72, 76, 52, 76])
)

demanda = np.array([560, 760, 720, 840])
capacidade = 40
necessidade = demanda / capacidade
desequilibrio = frota_atual - necessidade

T = np.outer(necessidade / necessidade.sum(), np.ones(4))
nova_frota = T @ frota_atual

regioes = ["Norte", "Sul", "Leste", "Oeste"]

print("Frota atual:")
for r, q in zip(regioes, frota_atual):
    print(f"{r}: {q:.0f} ônibus")

print("\nÔnibus necessários:")
for r, q in zip(regioes, necessidade):
    print(f"{r}: {q:.0f} ônibus")

print("\nDesequilíbrios (Frota Atual - Necessidade):", desequilibrio)

print("\nMatriz de Transformação T:")
print(np.round(T, 4))

print("\nNova distribuição:")
for r, q in zip(regioes, nova_frota):
    print(f"{r}: {q:.0f} ônibus")

print("\nSoma das colunas de T:", T.sum(axis=0))
print("Todos os coeficientes >= 0?", bool(np.all(T >= 0)))
print("Total de ônibus:", nova_frota.sum())
```
